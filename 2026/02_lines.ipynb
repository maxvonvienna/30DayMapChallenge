# Cycling through Vienna — one line at a time!

This map reveals the intricate geometry of Vienna’s cycling network, from the major cycling routes shown in dark blue to the smaller connecting routes in light blue. Together, they form a fascinating web of urban mobility, highlighting both the dense network in the city centre and the connections stretching towards the outskirts. Stripped of streets, buildings, and other distractions, the cycling infrastructure becomes a visual portrait of how Vienna connects its neighbourhoods on two wheels.

```python
import geopandas as gpd
import matplotlib.pyplot as plt
from pathlib import Path

MAP_PATH = Path.cwd() / "maps"
MAP_PATH.mkdir(exist_ok=True, parents=True)
```

```python
wfs_url = (
    "https://data.wien.gv.at/daten/geo?"
    "service=WFS&"
    "version=2.0.0&"
    "request=GetFeature&"
    "typeNames=ogdwien:RADNETZOGD&"
    #"bbox=16.378,48.203,16.385,48.206,EPSG:4326&"
    "srsName=EPSG:31256"
)

gdf = gpd.read_file(wfs_url)
```

```python
gdf.plot()
```

```python
fig, ax = plt.subplots(figsize=(10, 10))

gdf[gdf["RANG"] == "N"].plot(
    ax=ax,
    color="#8ECAE6",
    linewidth=0.5,
    label="N"
)

gdf[gdf["RANG"] == "H"].plot(
    ax=ax,
    color="#126782",
    linewidth=1.0,
    label="H",
)

ax.set_axis_off()
ax.set_aspect("equal")
plt.tight_layout()
plt.savefig(MAP_PATH / "map02.png", dpi=300, facecolor="white", bbox_inches="tight")
```
