**Commuting to Vienna – A City Beyond Its Borders**

Where does Vienna end, and where does its functional urban region begin? This map visualizes commuting flows from municipalities in Lower Austria and Burgenland towards Vienna, revealing the extensive network of daily interactions between the Austrian capital and its surrounding region.

Each line represents a commuting connection between a municipality and Vienna, transforming thousands of individual journeys into a dense web of spatial relationships. The overlapping lines highlight Vienna's role as a major employment centre and illustrate how strongly the city is connected to its hinterland. While administrative boundaries define where municipalities begin and end, commuting patterns reveal a different geography – one shaped by everyday mobility, economic interdependence, and functional relationships that extend far beyond the city limits.
