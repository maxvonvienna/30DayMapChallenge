I wanted to capture not only the geography of that time but also the visual character of historical cartography, with delicate linework, muted colors, and subtle textures. As someone who spends much of his time analyzing how cities and regions develop, I find it fascinating to look at the structures that existed long before today's urban landscape took shape. Sometimes, understanding a city means looking at what was there before.
