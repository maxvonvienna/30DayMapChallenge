### Day 20 – Hexagons | #30DayMapChallenge

**The Urban Fabric of Vienna – Population Meets Buildings**

What makes a place urban? Is it the number of people living there, the density of buildings, or the relationship between the two?

For today's challenge, I explored the spatial structure of the Vienna metropolitan region using a grid of 1,000-metre hexagons. Each hexagon combines two indicators: population density and building density. Instead of mapping these variables separately, I used a bivariate color scheme to reveal how they interact across the urban landscape.

The resulting map highlights the dense urban core of Vienna, where high population and building densities coincide. Beyond the city, the pattern becomes increasingly fragmented, revealing smaller settlement concentrations, suburban development, and the transition towards more sparsely populated rural areas. The different color combinations also distinguish places where building density and population density do not necessarily follow the same pattern.

Hexagons provide a consistent spatial framework for comparing these relationships across municipal boundaries. Their regular shape creates a continuous representation of settlement structures, making it easier to recognize clusters, gradients, and spatial discontinuities that administrative units might obscure.

**Urbanity is not defined by buildings or people alone, but by how they come together in space.**
