### Day 19 – Senses: Smell | #30DayMapChallenge

**The Smell of the Wind – Mapping an Invisible Landscape**

How do you map something you cannot see, but might be able to smell? Unlike roads, buildings, or administrative boundaries, odors are constantly moving through space, shaped by wind and atmospheric conditions.

For today's challenge, I explored the potential dispersion of odors originating from Vienna's main wastewater treatment plant in Simmering. Using historical hourly wind data from Open-Meteo and a simplified Gaussian plume model implemented in Python, I simulated how a hypothetical odor plume could spread across the surrounding area during November 2025. Each hourly plume was calculated on a 50-metre grid and aggregated to visualize how frequently different locations might fall within its influence.

The resulting map reveals a distinct eastward dispersion pattern, reflecting the wind conditions represented in the simulation. Colors indicate the percentage of modeled hours in which a relative plume intensity threshold is exceeded. The visualization illustrates how the potential spatial influence of a single source can extend beyond its immediate surroundings, creating an invisible geography shaped by the atmosphere.

Importantly, this is an illustrative model rather than a measurement of actual odor pollution. It assumes constant emissions and simplified dispersion conditions, without accounting for terrain, atmospheric stability, or measured odor concentrations.

**The map makes an invisible process visible: while the source remains fixed, its potential area of influence is constantly changing with the wind.**
