### Day 25 – Is This a Map? | #30DayMapChallenge

**The Sound of Vienna – When Geography Becomes Music**

Can a city be transformed into music? And if a musical composition is derived entirely from geographic information, can we still call it a map?

For today's challenge, I decided to explore the boundaries between cartography, data visualization, and music. Instead of representing Vienna through streets, buildings, or administrative boundaries, I transformed its road network into an experimental musical score.

Using real street geometries from OpenStreetMap and Python, I translated geographic properties into musical parameters. Each note represents a street segment: its east-west position determines when it is played, its north-south position defines the pitch, and its length influences the duration. The resulting composition unfolds geographically from west to east, turning Vienna's spatial structure into a sequence of musical events.

The visualization consists of five musical staves, where thousands of street segments become notes, forming patterns of varying density, rhythm, and pitch. Some passages appear almost empty, while others develop into dense clusters of overlapping notes, reflecting the underlying distribution of streets across the city.

The same geographic transformation was also exported as a MIDI composition, allowing the map to be experienced not only visually but through sound. The score is intentionally experimental rather than conventional musical notation, but its structure remains directly connected to real geographic data.

**So, is this a map? Or is it music? Perhaps it is both.**

By translating space into sound, this experiment challenges the idea that maps must preserve geographic shapes or even be visual representations. Sometimes, exploring a city from a different perspective means changing not just how we see it, but how we hear it.
