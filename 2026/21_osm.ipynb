### Day 21 – OpenStreetMap | #30DayMapChallenge

**Mapping Everyday Commerce – The Retail Landscape of Vienna**

What can the distribution of shops tell us about the spatial structure of a city? From supermarkets and bakeries to clothing stores, electronics shops, and small neighborhood businesses, retail locations reflect how urban life is organized.

For today's challenge, I used OpenStreetMap data to explore the geography of retail activities across Vienna and its surrounding municipalities. Each point represents a mapped shop, classified according to its `shop=*` tag in OpenStreetMap. Different colors distinguish retail categories, revealing the diversity of commercial activities across the metropolitan region.

The resulting map highlights a strong concentration of shops in Vienna's densely built-up urban core, with additional clusters extending along major settlement corridors and into surrounding towns. Supermarkets, convenience stores, and other everyday retail services form a widespread network, while the overall density of commercial locations decreases towards the urban periphery.

Beyond the spatial patterns, this map demonstrates the value of OpenStreetMap as an open, community-driven geographic database. Its detailed classification of individual businesses makes it possible to explore urban functions at a remarkably fine spatial scale. At the same time, the distribution reflects what contributors have mapped, meaning that differences in data completeness may also influence the observed patterns.

**Every shop is a small piece of geographic information. Together, they reveal the commercial structure of an entire metropolitan region.**
