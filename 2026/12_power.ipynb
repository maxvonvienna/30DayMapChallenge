Solar Power Above Vienna

How much energy potential is hidden on Vienna's rooftops? This map explores the spatial distribution of photovoltaic potential across the city, highlighting rooftops suitable for solar energy generation. Based on Vienna's photovoltaic potential dataset, the analysis combines a digital surface model, solar radiation modelling, aerial imagery, and roof inclination to estimate the suitability and potential installed capacity of individual roof surfaces. The colors represent the estimated photovoltaic capacity in kilowatt-peak (kWp), revealing how solar potential varies across Vienna's urban fabric.

While most rooftops appear in subtle purple tones, larger buildings with substantial photovoltaic potential stand out in green and yellow. The map illustrates how the city's existing building stock could contribute to renewable energy production without requiring additional land. The values represent theoretical estimates rather than actual installed capacity, offering a spatial perspective on Vienna's untapped solar energy potential.

```python
import requests
import geojson
import geopandas as gpd
from matplotlib import pyplot as plt
```

```python
url = "https://data.wien.gv.at/daten/geo"
params = dict(
    service="WFS",
    version="2.0.0",
    request="GetFeature",
    typeNames="ogdwien:BEZIRKSGRENZEOGD",
    outputFormat="application/json",
    srsName="EPSG:31287"
)
r = requests.get(url, params=params)
geodata = geojson.loads(r.content)
gdf_bezirke = gpd.GeoDataFrame.from_features(geodata, crs="EPSG:31287")

params["typeNames"] = "ogdwien:SOLARANLAGLSTGOGD"
r = requests.get(url, params=params)
geodata = geojson.loads(r.content)
gdf_solar = gpd.GeoDataFrame.from_features(geodata, crs="EPSG:31287")

```

```python
# Schrift für den ganzen Plot – muss VOR dem Erzeugen der Figure gesetzt werden
plt.rcParams.update({
    "font.family": "Arial Narrow",
    "font.size": 11,
})

fix, ax = plt.subplots(figsize=(12, 12))
gdf_solar.plot(ax=ax, column="ANLAGENLEISTUNG", legend=True, vmin=100, vmax=1000, legend_kwds=dict(
    orientation="horizontal",
    shrink=0.45,
    pad=0.02,
    aspect=50,
    format="{x:.0f}",
))

ax.set_axis_off()
plt.tight_layout()
plt.savefig("maps/map12.png", dpi=300, bbox_inches="tight")
plt.show()
```
