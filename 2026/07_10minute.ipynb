#30DayMapChallenge – Day 7: 10 Minutes

Austria, connected by lines.

For Day 7 of the #30DayMapChallenge, the challenge was to create a map in just 10 minutes. I decided to keep things simple and visualize Austria's electricity transmission network. By reducing the map to its essential elements, the result is an almost abstract drawing of lines connecting different parts of the country. Without borders, cities, or labels, the infrastructure itself becomes the geography. What I particularly like about this map is how a complex technical network turns into something resembling a minimalist piece of art. Sometimes, cartography is not about adding more information, but about knowing what to leave out.
