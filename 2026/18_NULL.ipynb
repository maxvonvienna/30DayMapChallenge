**Day 18 – NULL | #30DayMapChallenge**

**Nothing to Map**

What happens when a map contains absolutely nothing? No roads, no buildings, no boundaries, no data. Just an empty canvas and a north arrow.

For today's theme, NULL, I decided to take cartographic minimalism to its logical extreme. Instead of visualizing missing values, empty spaces, or gaps in geographic information, I removed the geography altogether. The only remaining element is a north arrow, suggesting that even when there is nothing to see, orientation still exists.

The result raises a simple question: **Is a map without geographic information still a map?**

Sometimes, the most radical cartographic decision is not what to include, but what to leave out.
