**The Hidden Backbone of Vienna – Mapping Network Centrality**

Not all streets are equally important. While some primarily provide local access, others play a crucial role in connecting different parts of a city. But how can we identify these connections without relying on road classifications or traffic counts?

For today's challenge, I explored the underlying structure of Vienna's road network using **Edge Betweenness Centrality**, a graph-theoretical measure that describes how frequently a connection appears along shortest paths between pairs of nodes.

Using OpenStreetMap road data and Python, I constructed a spatial network and analyzed 35,000 sampled origin-destination paths. Shortest paths were calculated using actual road lengths, allowing me to estimate the structural importance of individual street segments. The resulting centrality values were then translated into a continuous color gradient, highlighting the most important connections in orange, red, and dark purple, while the remaining network fades into the background.

The map reveals a distinct backbone of highly connected routes extending across Vienna. Major corridors, bridges, and connections between different parts of the city emerge as prominent elements, illustrating how a relatively small subset of streets contributes disproportionately to the network's overall connectivity.

Interestingly, this is not a map of traffic volumes. A street with high betweenness centrality is not necessarily the busiest, but it occupies a strategically important position within the network's topology. The analysis uses an undirected network and sampled shortest paths, making it an approximation of structural centrality rather than a representation of actual travel behavior.

**The result reveals a different geography of Vienna: one defined not by where streets are located, but by how they connect the city.**
